## 2. Moving Average (MA) of the various stocks

A moving average (MA) is used to smooth out price data by creating a constantly updated average price. The average is taken over a specific period of time, like 10 days, 20 minutes, 30 weeks, or any time period the trader chooses.

Types of Moving Averages:
1. Simple Moving Average (SMA) → Average closing price over a set period.
2. Exponential Moving Average (EMA) → Gives more weight to recent prices for a faster response to price changes.


How to Interpret Moving Averages?

✅ Price Above MA → Uptrend (Bullish Signal) 

✅ Price Below MA → Downtrend (Bearish Signal)

✅ Short-Term MA Crosses Above Long-Term MA → Golden Cross (Bullish)

✅ Short-Term MA Crosses Below Long-Term MA → Death Cross (Bearish)

```python
# Define moving average periods
ma_day = [10, 20, 50]

for ma in ma_day:
    for company_name,df in stock_data.items():
        column_name = f"MA for {ma} days"
        df[column_name] = df['Close'].rolling(ma).mean()
        

fig, axes = plt.subplots(nrows=2, ncols=2)
fig.set_figheight(10)
fig.set_figwidth(15)


stock_data['AAPL'][['Close', 'MA for 10 days', 'MA for 20 days', 'MA for 50 days']].plot(ax=axes[0,0])
axes[0,0].set_title('APPLE')

stock_data['GOOG'][['Close', 'MA for 10 days', 'MA for 20 days', 'MA for 50 days']].plot(ax=axes[0,1])
axes[0,1].set_title('GOOGLE')

stock_data['MSFT'][['Close', 'MA for 10 days', 'MA for 20 days', 'MA for 50 days']].plot(ax=axes[1,0])
axes[1,0].set_title('MICROSOFT')

stock_data['AMZN'][['Close', 'MA for 10 days', 'MA for 20 days', 'MA for 50 days']].plot(ax=axes[1,1])
axes[1,1].set_title('AMAZON')


fig.tight_layout()
plt.show()
```

## 3. Average Daily Return of the Stock

Now that we've explored some of the basics, let’s take a step further and look into the risk side of things. Instead of just focusing on the overall stock price, we’ll zoom in on how it changes from day to day. These daily returns can tell us a lot about how volatile a stock is — and that’s a key part of understanding risk.

To start, we’ll use Apple’s stock data and calculate its daily percentage change using Pandas. This will give us a clearer picture of how the stock moves on a regular basis.

```python
# We'll use pct_change to find the percent change for each day
for company_name,df in stock_data.items():
    df['Daily Return'] = df['Close'].pct_change()  #calculates the percentage change from the previous day’s closing price

# Then we'll plot the daily return percentage
fig, axes = plt.subplots(nrows=2, ncols=2)
fig.set_figheight(10)
fig.set_figwidth(15)

stock_data['AAPL']['Daily Return'].plot(ax=axes[0,0], legend=True, linestyle='--', marker='o')
axes[0,0].set_title('APPLE')

stock_data['GOOG']['Daily Return'].plot(ax=axes[0,1], legend=True, linestyle='--', marker='o')
axes[0,1].set_title('GOOGLE')

stock_data['MSFT']['Daily Return'].plot(ax=axes[1,0], legend=True, linestyle='--', marker='o')
axes[1,0].set_title('MICROSOFT')

stock_data['AMZN']['Daily Return'].plot(ax=axes[1,1], legend=True, linestyle='--', marker='o')
axes[1,1].set_title('AMAZON')

fig.tight_layout()
plt.show()
```

2. Google had a massive upward spike around early May 2024 — something really positive happened. But otherwise, it seems more stable with smaller swings compared to Apple.

## 4. Correlation between different Stock Closing Prices

Correlation tells us how two things move together — like whether two stocks usually go up or down at the same time. The value of correlation is always between -1 and +1:

+1 means they move exactly the same way.

-1 means they move in completely opposite directions.

0 means there’s no connection at all between how they move.

But remember, correlation doesn’t mean one causes the other — it just shows a relationship, not the reason behind it.

Now, if we want to compare how all the stocks in our list behave, we’ll gather the 'Close' prices from each stock into one table (called a DataFrame). This way, we can look at their daily returns side by side and see how they are related.

```python
# Fetching all the Close prices for the stocks into one dataframe 
closing_df = yf.download(tech_list, start=start, end=end)['Close']

# Compute daily percentage returns
tech_rets = closing_df.pct_change()
tech_rets.head()
```

## 6. Forecasting APPLE Inc's closing stock price

```python
# Get the stock quote for Apple (AAPL)
df = yf.download('AAPL', start='2012-01-01', end=datetime.now().strftime('%Y-%m-%d'))

# Show the data
df
```

```python
plt.figure(figsize=(16,6))
plt.title('Close Price History')
plt.plot(df['Close'])
plt.xlabel('Date', fontsize=18)
plt.ylabel('Close Price USD ($)', fontsize=18)
plt.show()
```

In this graph we can see that in early 2025, there is a sharp and steep drop in Apple's closing price. It appears to be the biggest single dip in 13 year timeline (since 2012). The price fell from above USD 250 to nearly USD 200, which is significant correction - more than 20%. It is still a strong long-term performer, but watch out for current market risks.

```python
# Create a new dataframe with only the 'Close column 
data = df.filter(['Close'])
# Convert the dataframe to a numpy array
dataset = data.values
# Get the number of rows to train the model on
training_data_len = int(np.ceil( len(dataset) * .95 ))

training_data_len
```

```python
# Drop the MultiIndex level ('Ticker') if needed
df.columns = df.columns.droplevel(1)  

# Ensure 'Close' is selected properly
data = df[['Close']]

# Convert to NumPy array (ensure it is 2D)
dataset = data.values.reshape(-1, 1)

# Validate dataset shape
print("Dataset Shape:", dataset.shape)  # Should be (rows, 1)

# Scaling
scaler = MinMaxScaler(feature_range=(0,1))
scaled_data = scaler.fit_transform(dataset)

# Print first few values
print(scaled_data[:5])
```

```python
# Create the training data set 
# Create the scaled training data set
train_data = scaled_data[0:int(training_data_len), :]
# Split the data into x_train and y_train data sets
x_train = []
y_train = []

for i in range(60, len(train_data)):
    x_train.append(train_data[i-60:i, 0]) # previous 60 prices
    y_train.append(train_data[i, 0])      # the "label" or target price
    if i<= 61:
        print(x_train)
        print(y_train)
        print()
        
# Convert the x_train and y_train to numpy arrays 
x_train, y_train = np.array(x_train), np.array(y_train)

# Reshape the data
x_train = np.reshape(x_train, (x_train.shape[0], x_train.shape[1], 1))
# x_train.shape
```

```python
from keras.models import Sequential
from keras.layers import Dense, LSTM

# Build the LSTM model
model = Sequential()
model.add(LSTM(128, return_sequences=True, input_shape= (x_train.shape[1], 1)))
model.add(LSTM(64, return_sequences=False))
model.add(Dense(25))
model.add(Dense(1))

# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error')

# Train the model
model.fit(x_train, y_train, batch_size=1, epochs=1)
```
